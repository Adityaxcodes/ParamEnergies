```python
# Import Required Library
from transformers import pipeline

# Load Pretrained BERT QA Pipeline
qa_pipeline = pipeline(
    "question-answering",
    model="bert-large-uncased-whole-word-masking-finetuned-squad",
)

# Context for Question Answering
context = """The Nile River is the longest river in the world, flowing through northeastern Africa for over 6,600 kilometers. It is a vital water source for Egypt and Sudan.  The river has two major tributaries: the White Nile and the Blue Nile. The White Nile is considered to be the headwaters and primary stream of the Nile itself...."""

# Define Question
question = "What is the longest river in the world?"

# Get Answer from Model
result = qa_pipeline(question=question, context=context)

# Print Answer and Confidence Score
print(f"Answer: {result['answer']}")
print(f"Score: {result['score']:.4f}")
```

```python
# Import Required Library
from transformers import pipeline

# Load Pretrained Sentiment Analysis Pipeline
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment",
)

# Texts to Analyze
texts = [
    "I absolutely loved this movie! It was fantastic.",
    "The product quality is terrible and the customer service is worse.",
    "It's okay, not the best but not the worst either.",
]

# Perform Sentiment Analysis
for text in texts:
    result = sentiment_pipeline(text)[0]
    print(f"Text: {text}")
    print(f"Sentiment: {result['label']} (Confidence: {result['score']:.2f})\n")
```

```python
# Import Required Library
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample Document Corpus
documents = [
    "The Eiffel Tower is in Paris.",
    "The Great Wall of China is visible from space.",
    "The Statue of Liberty is in New York.",
    "Paris is the capital of France.",
    "Space exploration is fascinating.",
]

# Preprocessing & Vectorization (TF-IDF)
vectorizer = TfidfVectorizer(stop_words="english")
doc_vectors = vectorizer.fit_transform(documents)

# Define a User Query
query = "Where is the Eiffel Tower?"

# Preprocess & Vectorize Query
query_vector = vectorizer.transform([query])

# Compute Cosine Similarity
similarities = cosine_similarity(query_vector, doc_vectors).flatten()

# Rank Documents by Similarity
ranked_indices = similarities.argsort()[::-1]

# Display Top Results
print("Top results for the query:\n")
for idx in ranked_indices[:3]:
    print(f"Score: {similarities[idx]:.4f} | Document: {documents[idx]}")
```
